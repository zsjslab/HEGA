```python
# 原始问答对格式修改
import pandas as pd
import json
import os
 #直接读取 Excel 文件，指定使用 openpyxl 引擎term_basev2_1203/KG_RAG_WORK/src/GraphRAG/dataset/xlsx_QA/question_0327.xlsx
xlsx_name = './xlsx_QA/question_0327.xlsx'
df = pd.read_excel(xlsx_name, engine='openpyxl')
# 打印数据
print(df)

# 从数据中提取问题描述列和标准答案列
data = df[['问题', '标准答案']]

# 将数据转换为指定的 JSON 格式
json_data = []
for _, row in data.iterrows():
    question = row['问题']
    answer = row['标准答案']
    json_data.append({
        'question': question,
        'answer': answer
    })
    
# 提取文件名（不带扩展名）
base_name = os.path.splitext(os.path.basename(xlsx_name))[0]  # 提取 'question_0327'
# 构造 JSON 文件路径
json_path = f'./TERM_BASE_QA/{base_name}.json'
# 将结果保存为 JSON 文件
with open(json_path, 'w', encoding='utf-8') as f:
    json.dump(json_data, f, ensure_ascii=False, indent=2)
```
